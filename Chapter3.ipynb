Multiple input nn

```python
def w_sum(a,b):
  assert(len(a)==len(b))
  output = 0
  for i in range(len(a)):
    output += a[i]*b[i]

  return output
weights = [0.1,0.2,0]
def neural_network(input,weights):
  prediction = w_sum(input,weights)
  return prediction

toes = [8.5,9.5,9.9,9.0]
wlrec = [0.65,0.8,0.8,0.9]
nfans = [1.2,1.3,0.5,1.0]

input = [toes[0],wlrec[0],nfans[0]]
pred = neural_network(input,weights)
print(pred)

```

NumPy Version

```python
import numpy as np

def w_sum(a,b):
  assert(len(a)==len(b))
  output = 0
  for i in range(len(a)):
    output += a[i]*b[i]

  return output
weights = np.array([0.1,0.2,0])
def neural_network(input,weights):
  prediction = input.dot(weights)
  return prediction

toes = [8.5,9.5,9.9,9.0]
wlrec = [0.65,0.8,0.8,0.9]
nfans = [1.2,1.3,0.5,1.0]

input = np.array([toes[0],wlrec[0],nfans[0]])
pred = neural_network(input,weights)
print(pred)

```

Multiple outputs

```python
def elementwise_multiplication(a,b):
  output = [0,0,0]
  assert(len(b)==len(output))
  for i in range(len(output)):
    output[i] = a*b[i]

  return output
weights = [0.3, 0.2, 0.9]
def neural_network(input,weights):
  prediction = elementwise_multiplication(input,weights)
  return prediction

toes = [8.5,9.5,9.9,9.0]
wlrec = [0.65,0.8,0.8,0.9]
nfans = [1.2,1.3,0.5,1.0]

input = wlrec[0]
pred = neural_network(input,weights)
print(pred)
```

Multiple Input and Outputs

```python


            #toes %win #fans
weights = [ [0.1, 0.1, -0.3], #hurt?
            [0.1, 0.2, 0.0], #win?
            [0.0, 1.3, 0.1] ] #sad?

def vec_to_matrix_multiplication(a,b):
  output = [0,0,0]

  assert (len(a)==len(b))

  for i in range(len(b)):
    for j in range(len(a)):
      output[i] += a[j]*b[i][j]
  return output

def neural_network(input, weights):
  layer = vec_to_matrix_multiplication(input,weights)
  return (layer)

toes =  [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65,0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

input = [toes[0],wlrec[0],nfans[0]]

print(neural_network(input,weights))
```

```python
            #toes %win #fans
ih_wgt = [ [0.1, 0.2, -0.1], #hid[0]
           [-0.1,0.1, 0.9], #hid[1]
           [0.1, 0.4, 0.1] ] #hid[2]

           #hid[0] hid[1] hid[2]
hp_wgt = [ [0.3, 1.1, -0.3], #hurt?
           [0.1, 0.2, 0.0], #win?
           [0.0, 1.3, 0.1] ] #sad?

weights = [ih_wgt, hp_wgt]

def neural_network(input, weights):
    hid = vec_to_matrix_multiplication(input,weights[0])
    pred = vec_to_matrix_multiplication(hid,weights[1])
    return pred

toes =  [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65,0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

# Input corresponds to every entry
# for the first game of the season.

input = [toes[0],wlrec[0],nfans[0]]
pred = neural_network(input,weights)

print(pred)
```
