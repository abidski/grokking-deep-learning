Gradient descent with multiple inputs

```python
toes =  [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

win_or_lose_binary = [1, 1, 0, 1]

input = [toes[0],wlrec[0],nfans[0]]
weights = [0.1, 0.2, -.1]
true = win_or_lose_binary[0]

def w_sum(a,b):
  assert (len(a)==len(b))
  output = 0
  for i in range(len(a)):
    output += a[i] * b[i]
  return output

def ele_multiplication(vector, scalar):
  delta_weight = [0,0,0]
  for i in range(len(vector)):
    delta_weight[i] = scalar*vector[i]

  return delta_weight

def neural_network(input, weights):

  prediction = w_sum(input, weights)
  return prediction

pred = neural_network(input, weights)

error = (pred - true)**2
alpha = 0.01
delta = pred -true
delta_weight = ele_multiplication(input, delta)
for i in range(len(weights)):
  weights[i] -= delta_weight[i] * alpha

print("Weights:" + str(weights))
print("Weight Deltas:" + str(delta_weight))

```
